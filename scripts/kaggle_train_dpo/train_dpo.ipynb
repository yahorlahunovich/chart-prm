# Reproducible Step-DPO Kaggle smoke run

This launcher checks out the recorded project revision and runs the canonical training notebook in three-step smoke mode on a T4.

```python
import os
from pathlib import Path
import subprocess

REVISION = '59910789b56d0fb3c512ffdd71680bf19e0d8639'
repo_dir = Path('/tmp/prm_project')
subprocess.run(
    ['git', 'clone', 'https://github.com/yahorlahunovich/prm_project.git', str(repo_dir)],
    check=True,
)
subprocess.run(['git', '-C', str(repo_dir), 'checkout', '--detach', REVISION], check=True)
os.environ['PRM_REPO_REVISION'] = REVISION
os.environ['DPO_SMOKE_ONLY'] = '1'
os.chdir(repo_dir)
print(f'Running Step-DPO smoke test from revision {REVISION}')
get_ipython().run_line_magic('run', 'notebooks/train_dpo.ipynb')
```
